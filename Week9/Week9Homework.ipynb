```python
df['Borough'] = df['Community_District'].astype(str).str[0]

## I have to define boroughs by community district to sort, because I removed borough rows, dividing each chunk of community districts. 
df['Borough'] = df['Borough'].map({
    '1': 'Manhattan',
    '2': 'Bronx',
    '3': 'Brooklyn',
    '4': 'Queens',
    '5': 'Staten Island'
})
```

```python
manhattan = df[df['Borough'] == 'Manhattan']
```

```python
manhattan.head()
```

```python
df['Borough'].value_counts()
```

```python
manhattan.info()
```

```python
manhattan.describe()
```

```python
manhattan.plot(
    x='Community_District',
    y=['Male', 'Female'],
    kind='bar',
    figsize=(12, 6)
)

plt.title('Male and Female Death Counts by Manhattan Community District (2022)')
plt.xlabel('Manhattan Community District')
plt.ylabel('Population Count')
plt.xticks(rotation=45)
plt.legend(title='Gender')
plt.tight_layout()
plt.show()
```
