```python
import pandas as pd
import plotly.express as px

deathrate_df = pd.read_csv("Deathratebycommunitydistrict.csv")

deathrate_df.head()
```

```python
print(deathrate_df.columns)
```

```python
deathrate_df.info()
```

```python
fig = px.bar(
    deathrate_df,
    x='Death_Rate',
    y='Community_District',
    orientation='h',
    title='Death Rates by NYC Community District',
    labels={
        'Death_rate': 'Death Rate',
        'Community_District': 'Community District'
    },
    hover_data=deathrate_df.columns.tolist()
)

fig.update_layout(
    yaxis={'categoryorder': 'total ascending'},
    template='plotly_white',
    height=900,
    xaxis_title='Death Rate',
    yaxis_title='NYC Community District'
)

fig.show()
```

```python
fig = px.scatter(
    deathrate_df,
    x='Community_District',
    y='Death_Rate',
    color='Death_Rate',
    size='Death_Rate',
    hover_data=deathrate_df.columns.tolist(),
    title='Death Rates Across NYC Community Districts',
    color_continuous_scale='Viridis'
)

fig.update_layout(
    template='plotly_white',
    xaxis_tickangle=-45
)

fig.show()
```

```python
fig = px.histogram(
    deathrate_df,
    x='Death_Rate',
    nbins=15,
    title='Distribution of Death Rates by Community District',
    labels={'Death_Rate': 'Death Rate'},
    color_discrete_sequence=['steelblue']
)

fig.update_layout(
    template='plotly_white',
    yaxis_title='Number of Districts'
)

fig.show()
```

```python
fig = px.box(
    deathrate_df,
    y='Death_Rate',
    points='all',
    hover_data=deathrate_df.columns.tolist(),
    title='Death Rate Distribution Across NYC Community Districts'
)

fig.update_layout(
    template='plotly_white',
    yaxis_title='Death Rate'
)
```

```python
fig = px.treemap(
    deathrate_df,
    path=['Community_District'],
    values='Death_Rate',
    color='Death_Rate',
    color_continuous_scale='Reds',
    title='Death Rates by NYC Community District'
)

fig.show()
```

```python
fig = px.density_heatmap(
    deathrate_df,
    x='Community_District',
    y='Death_Rate',
    nbinsy=10,
    title='Death Rate Density Across Community Districts',
    color_continuous_scale='YlOrRd'
)

fig.update_layout(
    template='plotly_white',
    xaxis_tickangle=-45
    )

fig.show()
```
