MoMA's artists were born across hundreds of years. Which birth years show up the most often in the collection?
Write code that:
Loads the whole dataset (you know how to do this already).
Builds a dictionary that counts how many artists share each BeginDate value — one key per year, one count per key.
Sorts that dictionary by count, largest first, and keeps just the top 10.
Prints each of those years next to its count.
Everything you need has already been covered somewhere in this course. For step 2, think about the if / else and in patterns from Part 5 and Part 1 — for each artist, is their birth year already a key in your dictionary or not? For step 3, sorted() takes a lambda key, which we covered in week 3.

```python
import csv

birth_years = {} #creating my  key-value pair and assigning birth as my key to get the value of the number born in birth year

with open("moma_artists.csv", "r", encoding="utf-8") as file: #opening CSV file in read mode
    reader = csv.DictReader(file) # setting it up to read each row as a dictionary
    
    for artist in reader:
        year = artist["BeginDate"] #pulling birth year 

        if year in birth_years: #checking if birth year exists in another value
            birth_years[year] += 1 #adding to the count if year exists already for another artist
        else: #runs if not in dictionary
            birth_years[year] = 1 #adds the year to dictionary 

top_10 = sorted(
    birth_years.items(), # pulling key-value pair
    key=lambda item: item[1], # sorting by the second element (count)  
    reverse=True # sorting from largest to smallest count of the same birth year
)[:10] #value here is the number I want to order

for year, count in top_10:
    print(year, count)
```
