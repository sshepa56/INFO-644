**Final Project Proposal: Mapping Mortality using NYC Open Data**

**Project Themes** 
This project will use NYC Open Data and other publicly available datasets to explore geographic patterns in mortality and their potential relationship to health care access and socioeconomic conditions across New York City.
I plan to visualize mortality trends by cause, year, and location, then compare those patterns with indicators such as health insurance coverage, poverty, and socioeconomic status (SES). My goal is to investigate whether certain communities experience disproportionately high mortality rates or burdens from particular illnesses, and whether these patterns coincide with barriers to health care access.
I am particularly interested in respiratory illness hospitalizations because they provide an opportunity to examine how a key morbidity marker varies across zip codes and over time.  
*Research question is: How do mortality patterns across NYC relate to differences in health care access and socioeconomic conditions?*
