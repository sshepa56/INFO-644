First, we bring in the moma_artists.csv

```python
import csv #brings in the csv functions

# the open function takes these 3 arguments:
# the file name/path
# "r" for read
# and the encoding type
with open("moma_artists.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file) #read the file as a dictionary object
    data = list(reader) #convert that dictionary object into a usable list called "data"

print(data[0]) # prints the first dictionary item
    
```

Simple Search of the data by first name

```python
name_of_artist = data[0]['DisplayName']
print(name_of_artist)
```

```python
first_name_of_artist = name_of_artist.split(" ") [0]
print(first_name_of_artist)
```

Generating a list of all artists with the first name mary:

```python
# step 1: make an empty list to hold artists with the name mary

artists_named_mary = [] #square brackets with no contents means an empty list

#step 2: create a for loop that goes through every item in the dataset

for artist in data:
    # step 3 extract the firist name
    first_name = artist['DisplayName'].split(" ")[0]
    # check if the first name matched "Mary"
    if first_name == "Mary":
       # add out Mary to the new list
        artists_named_mary.append(artist)
print(len(artists_named_mary))

for mary in artists_named_mary[:10]:
## this means the first 10 in the list
    print(mary['DisplayName'])
```

Sorting & Resorting Dictionaries

example: sorting by birth year (BeginDate)

## if you are reordering do not change og dataset, create a copy to modify for each modified version

```python
artists_by_year = sorted(data, key=lambda artist: int(artist['BeginDate']))

# sorted takes two parameters
# first one is the dataset we call data here, which is the variable containing the complete MoMA dataset
# second one is the sorting logic, which is represented by a lambda function
# here, that lambda function is saying "for each artist in the olist, extract the BeginDate and convert it to an int'

print(artists_by_year[-1])
```

Filtering Datasets - filtering out entries that have year=0

```python
artists_with_year = []

for artist in data:
    if artist['BeginDate'] != "0": # != means not equal
        artists_with_year.append(artist)
artists_by_year_filtered = sorted(artists_with_year, key=lambda artist: int(artist['BeginDate']))
                                  
print(artists_by_year_filtered[0])
```

```python
for artist in artists_by_year_filtered[:20]:
    print(artist['DisplayName'])
```
