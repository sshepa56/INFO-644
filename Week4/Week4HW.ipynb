## CSV Practice

```python
import csv

with open("moma_artists.csv") as file:
    artists = list(csv.DictReader(file))

artists_with_gender_data = []

### creating a variable for gender to allow for filtering

for artist in artists:
    if artist["Gender"]:
        artists_with_gender_data.append(artist)

### Using loops to filter artists by gender
male_artists = []
female_artists = []
non_binary_artists = []
other_artists = []

for artist in artists_with_gender_data:
    if artist["Gender"] == "male":
        male_artists.append(artist)
    elif artist["Gender"] == "female":
        female_artists.append(artist) 
    else:
        other_artists.append(artist)
### using len to produce the counts for my gender variables

print(f"Male artists: {len(male_artists)}")
print(f"Female artists: {len(female_artists)}")
print(f"Non-binary artists: {len(non_binary_artists)}")
```
